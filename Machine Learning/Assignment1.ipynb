```python
def average(values):
    total=0
    for i in range (len(values)):
        total+=values[i]
    return total/len(values)
```

## Explanation of `describe()` Output

- count: Number of non-missing values in each column.
- mean: Average value of the column.
- std: Standard deviation, which measures how spread out the values are from the mean.
- min: Smallest value in the column.
- 25%: First quartile (25% of the data is below this value).
- 50%: Median (middle value of the data).
- 75%: Third quartile (75% of the data is below this value).
- ax: Largest value in the column.

```python
df.loc[df["name"] == "Rohan","test1"] = 400
```

```python
print("New Mean of Test 1:", df["test1"].mean())
```

```python
print("New Standard Deviation of Test 1:", df["test1"].std())
```

After changing Rohan's Test 1 marks from 40 to 400, the mean increased significantly from about 69.33 to 99.33. The standard deviation also increased sharply because the incorrect value is an extreme outlier, making the data much more spread out. This demonstrates that both the mean and **standard deviation** are highly sensitive to outliers and data entry errors.
